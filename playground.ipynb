```python
from regmem import ConsolidationEngine, Consolidator, RegressionGenerator, Verifier
from regmem.llm import LLM
from regmem.typedefs import MemoryState, RegressionSuite
```

```python
llm = LLM("Qwen/Qwen3-4B")
engine = ConsolidationEngine(
    Consolidator(llm), RegressionGenerator(llm), Verifier(llm),
    max_repairs=2,
)
```

```python
# Rerun this cell to start a fresh experiment.
memory, suite = MemoryState(), RegressionSuite()
evidence_batches = [
    "I prefer tea without milk.",
    "For breakfast I prefer coffee, but still without milk.",
]
```

```python
for evidence in evidence_batches:
    result = engine.consolidate(memory, suite, evidence, save_run=True)
    if result.accepted:
        memory, suite = result.memory, result.suite

    print(f"Accepted: {result.accepted} | Attempts: {result.attempts}")
    print("Memory:", [item.content for item in memory.items])
    print("Tests:", len(suite.tests))
    print("Log:", result.run_path)
    print()
```
